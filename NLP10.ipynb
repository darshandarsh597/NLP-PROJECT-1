```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns

import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

from datasets import load_dataset

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("haizelabs/banking77")
```

```python
train_df = pd.DataFrame(ds["train"])
test_df = pd.DataFrame(ds["test"])

print("Training data:", train_df.shape)
print("Testing data:", test_df.shape)

print(train_df.head())
```

```python
print(train_df.isnull().sum())
print(test_df.isnull().sum())
```

```python
print(train_df["label_gt"].value_counts().sort_index())
```

```python
train_df["category"] = train_df["label_gt"].astype(str)

test_df["category"] = test_df["label_gt"].astype(str)

print(train_df[["text", "category"]].head())
```

```python
stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    
  
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    text = re.sub(r'\s+', ' ', text).strip()
    
   
    words = text.split()
    words = [word for word in words if word not in stop_words]
    
    return " ".join(words)
```

```python
train_df["clean_text"] = train_df["text"].apply(clean_text)
test_df["clean_text"] = test_df["text"].apply(clean_text)
```

```python
category_counts = train_df["category"].value_counts()

plt.figure(figsize=(12, 7))

category_counts.head(15).plot(kind="bar")

plt.title("Top 15 Most Frequent Banking Complaints")
plt.xlabel("Banking Issue")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=75)

plt.tight_layout()
plt.show()
```

```python
from collections import Counter

all_words = " ".join(train_df["clean_text"]).split()

word_counts = Counter(all_words)

print(word_counts.most_common(20))
```

```python
words = [x[0] for x in word_counts.most_common(15)]
counts = [x[1] for x in word_counts.most_common(15)]

plt.figure(figsize=(10, 6))

plt.bar(words, counts)

plt.title("Most Frequent Words in Banking Complaints")
plt.xlabel("Words")
plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
X_text = train_df["clean_text"]
y = train_df["label_gt"]

X_test_text = test_df["clean_text"]
y_test = test_df["label_gt"]
```

```python
tfidf = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_text)

X_test_tfidf = tfidf.transform(X_test_text)

print("Training shape:", X_train_tfidf.shape)
print("Testing shape:", X_test_tfidf.shape)
```

```python
nb_model = MultinomialNB()

nb_model.fit(X_train_tfidf, y)

nb_pred = nb_model.predict(X_test_tfidf)

nb_accuracy = accuracy_score(y_test, nb_pred)

print("Naive Bayes Accuracy:",
      nb_accuracy)
```

```python
svm_model = LinearSVC()

svm_model.fit(X_train_tfidf, y)

svm_pred = svm_model.predict(X_test_tfidf)

svm_accuracy = accuracy_score(
    y_test,
    svm_pred
)

print("SVM Accuracy:",
      svm_accuracy
)
```

```python
lr_model = LogisticRegression(
    max_iter=1000
)

lr_model.fit(X_train_tfidf, y)

lr_pred = lr_model.predict(X_test_tfidf)

lr_accuracy = accuracy_score(
    y_test,
    lr_pred
)

print("Logistic Regression Accuracy:",
      lr_accuracy)
```

```python
results = pd.DataFrame({
    "Model": [
        "Naive Bayes",
        "Logistic Regression",
        "SVM"
    ],
    
    "Accuracy": [
        nb_accuracy,
        lr_accuracy,
        svm_accuracy
    ]
})

print(results)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["Accuracy"]
)

plt.title("Banking Complaint Classification Model Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")

plt.ylim(0, 1)

plt.show()
```
